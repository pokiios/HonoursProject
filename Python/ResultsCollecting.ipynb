# Dead of Night Results

## Importing Libraries + Initialisation

```python
# Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import neurokit2 as nk
```

```python
# Load CSV File into dataframe
file_path = 'CollectedData.csv'
df = pd.read_csv(file_path)
```

```python
# Participant information for easier table editing
participant_number = 0
```

```python
#Inspect Data
display(df.head())
print(f"Column Names: {df.columns.tolist()}")
```

```python
# Turn columns into variables
time_column = 'Timestamp'
RMSSD_column = 'RMSSD'
RSP_column = 'RSP'
```

## Graph Initialisation

```python
# Set style for graph
sns.set_style("whitegrid")
plt.rcParams.update({'font.size': 12})
```

```python
# Create figure for the plot
plt.figure(figsize=(10, 6))
```

## RMSSD/RSP Over Time

```python
# Plotting RMSSD over time
ax1 = plt.gca()
ax1.plot(df['Timestamp'], df['RMSSD'], 'r-', linewidth=1.5)
ax1.scatter(df['Timestamp'], df['RMSSD'], s=30, color='red', alpha=0.6)
ax1.set_xlabel('Time (seconds)', fontsize=14)

#Plotting RSP over time
ax2 = ax1.twinx()
ax2.plot(df['Timestamp'], df['RSP'], 'b-', linewidth=1.5)
ax2.scatter(df['Timestamp'], df['RSP'], s=30, color='blue', alpha=0.6)

# Collating points
lines1, lables1 = ax1.get_legend_handles_labels()
lines2, lables2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, lables1 + lables2, loc='upper right')
plt.title(f'Participant {participant_number}: RMSSD + RSP over Time', fontsize = 16)

plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
